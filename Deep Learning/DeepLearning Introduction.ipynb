## In Class Notes

- Deep Learning: Is a technique that mimics the behavior of a neuron and is primarily devised to achieve the targets of machine learning.
   - Subset of machine learning that uses algorithms inspired by the structure and the function of the brain's neural network.
   - Since deep learning algorithms are inspired by the neural networks in a brain, the algorithms are generally referred to as Artificial Neural Networks.
   - Why Deep Learning?
      - Can be used to breakdown inputs to first understand low level features (lines and edges) and then progress to understand mid-level features (shapes and objects) and finally get a real world understanding of the input by combining features from lower levels(cars or faces).
      - The key idea of deep learning is to learn the underlying features and create a process to observe and distinguish the features after observing the input data. 
- Perceptron: The structural building block of Deep Learning. A single Neuron.
   - 
   - Inputs: The values passed to the neuron. They are always numerical. Denoted by $x_1,x_2,...,x_m$
   - Weights: Numbers that are multiplied to the inputs. They are initialized randomly. Denoted by $w_1,w_2,...w_m$
   - Bias: A constant error term, which is also numerical in nature, added to the summed product of weights and input. Denoted by $w_0$.
      - Helps shift the output before passing it on to the activation function.
   - Activation Function: Non Linear in nature. These functions accept a single numerical value as input and perform a mathematic function on the given input. Denoted as $g$
   - Forward Propagation:
      - Process:
         1. The input and weights are multiplied.
         2. Then each product is summed up and combined to a single value.
         3. Then the value is passed through a non-linear activation function.
      - Mathematical Illustration:
         $$ \hat{y} = g(w_0 + \sum_{i=1}^{m} X_i w_i) $$

   - Vector Product (Dot Product): The fundamental calculation for a single neuron is a vector product (specifically, the dot product) between the input vector from the previous layer's activations and the weight vector specific to that neuron.
      - Operation: The dot product involves multiplying corresponding elements of two vectors and summing the results to produce a single scalar value.
      - Context: For a single input data point x and a set of weights w for one neuron, the operation is $z=\mathbf{x}\cdot \mathbf{w}+w_0$, where $w_0$ is the bias. This weighted sum, $z$, is then passed through an activation function.
   - Matrix Multiplication: Matrix multiplication is a generalization of the dot product used to perform these calculations simultaneously for all neurons in a layer and for a batch of multiple input data points (samples).
      - Operation: A matrix multiplication between matrix $X$ (e.g., input data) and matrix $W$$ (e.g., weights) computes all possible pairwise dot products between the rows of \$X$ and the columns of $W$ to produce a new matrix $C$.
      - Context: In a neural network layer, the operation is represented as: $Z=XW+B$ X is the input matrix (each row is a data sample).W is the weight matrix (containing the connection weights).B is the bias vector (added to each row of the result).Z is the resulting matrix of weighted sums for the layer, which then has an activation function applied element-wise. 

   - Clean Mathematical Illustration:
      $$ \hat{y} = g(w_0 + \mathbf{X^T} \mathbf{W}) $$
   - Importance of Activation Functions:
      - The purpose of an activation function is to introduce non-linearity into the network.
      - Linear activation functions produce linear results no matter the network size.
         - A feed-forward neural network with linear activation and any number of hidden layers is equivalent to just a linear neural neural network with no hidden layer. For example lets consider the neural network in figure with two hidden layers and no activation.

      - Non linear activation function allow us to approximate arbitrarily complex functions.
         - We need non-linear activation functions to approximate non-linear functions and most real world problems are highly complex and non-linear. In fact when the activation function is non-linear, then a two-layer neural network with sufficiently large number of hidden units can be proven to be a universal function approximator.

   - Multi Output Perceptron:
      - Perceptron networks that have multiple outputs.
      - 
      - Dense Layer:  is a layer of neurons in the network where each neuron in the layer receives input from all neurons in the preceding layer.
         - Dense layers are commonly used as hidden layers in Multi-Layer Perceptrons (MLPs), or in the end stages of other architectures like Convolutional Neural Networks (CNNs) for classification based on the features extracted by earlier layers.
   
- Artificial Neural Network (ANN): 
   - A neural network (also called an artificial neural network or ANN) is an adaptive system that learns by using interconnected nodes or neurons in a layered structure that resembles a human brain. A neural network can learn from data, so it can be trained to recognize patterns, classify data, and forecast future events. A neural network breaks down the input into layers of abstraction. It can be trained using many examples to recognize patterns in speech or images just as the human brain does. The neural network behavior is defined by the way its individual elements are connected and by the strength, or weights, of those connections. These weights are automatically adjusted during training according to a specified learning rule until the artificial neural network performs the desired task correctly.
   - A powerful algorithm that mimics the neuron in the brain.
   - Synonyms: Neural network, Model, Net.

   - Single Layer Neural Network: Is a network where there is a hidden layer of neurons placed between the input layer and the output layer.
      - 
      - Hidden Layer: The layer which accepts the inputs from previous layer, computes the values(forward propagation), and passes it to the next layer. These layers are not observed and it appears to be hidden when compared to the input and output layer. It learns over the iterations of data passed through it.
      - Each neuron in the hidden layer performs the same operations, such as, multiplying the weights, adding bias and applying the non linear activation function, but the weights and bias for each neuron would be different.
   - Deep Neural Network: Is a network consisting of sequentially stacking hidden dense layers between the output and the input layer.
      - 
      - For more complex tasks, you would ideally want to introduce more depth by adding more layers, this in turn helps the network model a non linear function.

- Applying Neural Networks:
   - Loss: Loss refers to the cost incurred by the network by making predictions against the actual/expected results.
   - Quantifies how "bad" a prediction is. It measures the distance between the model's predictions and the actual "ground truth" answers. A smaller loss means predictions are closer to the truth
   - Empirical Loss: Measures the aggregate loss across the entire dataset.
      - AKA: Cost Function, Objective Function, Empirical Risk.
      - Types of Cost Function:
         - Binary Cross Entropy Loss: Used for models that output a probability between 0 and 1. (Classification)
         - Mean Squared Error Loss: Used with regression models that output real numbers.

- Training Neural Networks:
   - Aim: To find the weights that achieves the lowest loss.
   - Loss Optimization: Deep learning models are primarily optimized by minimizing a loss function through iterative updates of weights and biases. 
   - Gradient Descent Algorithm: The core algorithm; it calculates the gradient of the loss with respect to each parameter and moves parameters in the negative gradient direction.
      - Gradient Descent Variants: The fundamental technique is Stochastic Gradient Descent (SGD), which uses small data subsets (mini-batches) to speed up training and escape local minima.
      - Adaptive Optimizers: Algorithms like Adam (Adaptive Moment Estimation) and RMSprop are industry standards in 2026. They dynamically adjust learning rates for each parameter, handling sparse gradients and noisy data more effectively than standard SGD.
      - Momentum-based Methods: Techniques like Nesterov Accelerated Gradient (NAG) add "momentum" to updates, helping the optimizer navigate "saddle points" and accelerate convergence in flat regions of the loss landscape.
   - Backpropagation: An efficient way to compute gradients through the network, from output to input, using the chain rule.
   - Hyperparameter Tuning: Adjusting settings like learning rate, batch size, and epochs to control the optimization process.
   - Regularization: Techniques (like Dropout, L1/L2) added to prevent overfitting, improving generalization to new data.
      - Regularization is a crucial concept in deep learning to prevent overfitting, where a model performs well on training data but poorly on new, unseen data. Regularization methods add penalties or constraints during training, encouraging the model to be simpler and more generalizable.
      - Here are some common regularization techniques:
         - Penalty-Based Methods (Modifying the Loss Function):
            - L1 Regularization (Lasso): Adds the absolute value of the weights to the loss function. This can drive some weights to exactly zero, effectively performing feature selection by making certain features irrelevant.
            - L2 Regularization (Weight Decay): Adds the squared magnitude of the weights to the loss function. This forces weights towards zero (though rarely exactly zero), leading to simpler models that are less sensitive to specific features.
         - Architecture-Based Methods:
            - Dropout: During each training step, a random fraction of neurons are temporarily deactivated (set to zero). This prevents neurons from becoming overly reliant on each other and forces the network to learn more robust and redundant representations.
            - Batch Normalization: Normalizes the inputs of layers within a mini-batch. This stabilizes the training process and acts as a regularizer by adding a small amount of noise.
         - Data-Based Methods:
            - Data Augmentation: Artificially increases the size and variability of the training dataset by applying various transformations (e.g., rotations, flips, zooms) to existing data. This exposes the model to a wider range of examples and improves its robustness.
         - Training Process Methods:
            - Early Stopping: Monitoring the model's performance on a separate validation set during training. Training is halted when the performance on the validation set starts to degrade, even if the training loss is still decreasing. This prevents the model from memorizing the training data too much.

   - Deep Sequence Modelling
      - Sequential Data: Sequential data is information where the order of data points is crucial, meaning each element is dependent on the ones before and after it, creating a meaningful sequence, like words in a sentence, stock prices over time, DNA strands, or video frames. Unlike independent data, the arrangement itself carries significant context and patterns, requiring specialized models (like RNNs, LSTMs) for analysis in tasks like translation, speech recognition, or time-series forecasting.
      - Key Characteristics
         - Order Matters: The sequence's integrity is vital for understanding; changing the order changes the meaning.
         - Dependencies: Each data point relates to its neighbors, creating temporal or positional coherence.
         - Contextual: The arrangement provides context, structure, and meaning to the data.
      - Common Examples
         - Text: Words in a sentence, characters in a document.
         - Time Series: Stock prices, sensor readings, weather data.
         - Bioinformatics: DNA, RNA, or protein sequences.
         - Audio & Video: Speech, music, video frames.
         - User Behavior: A user's clickstream or purchase history.       
      - Applications of Sequence Modelling:
         - Many to One:
            - Input is sequential and the output is singular.
            - Ex: Sentiment Analysis.
         - One to Many:
            - Input is singular and the output is sequential.
            - Ex: Image to Text (Image captioning)
         - Many to Many:
            - Input and output are sequential.
            - Ex: Language Translation
      - Neurons with recurrence: recurrent neurons feed their own output back into themselves as an additional input for the next step.
         - 
         - Feedback Loop: is a connection that passes information from a neuron's output back as an input to itself (or the next neuron in the sequence) at the next time step, creating a "memory" that allows the network to use past inputs to influence current outputs, essential for processing sequential data like text or time series.
         - How Recurrent Neurons Work: In a recurrent setup, each neuron maintains a hidden state, which acts as a form of short-term memory. 
            - Dual Inputs: At any given time step ($t$), the neuron receives two inputs: the current external data ($x_{t}$) and its own hidden state from the previous time step ($h_{t-1}$).
            - Weighted Processing: The neuron applies specific weights to both the new input and the past memory. It then passes this combined value through an activation function (typically tanh or ReLU) to produce a new hidden state ($h_{t}$).
            - Memory Persistence: This new state is used to generate an output and is also saved to be fed back into the neuron for the next element in the sequence.
      - *Recurrent Neural Networks*: is a type of neural network designed to process sequential data (like text, speech, or time series) by using internal memory (a hidden state) to remember past information
         - RNNs have a state ($h_{t}$) that is updated at each time step as a sequence is processed.
         - Recurrence Relation:
            $$ h_{t} = f_{w}(x_{t}, h_{t-1})
         - 
         - Core Concepts
            - Sequential Memory: RNNs "unfold" over time steps. At each step, the network takes a current input and the hidden state (memory) from the previous step.
            - Shared Parameters: RNNs use the same weights and biases across all time steps, which minimizes the complexity of the model and allows it to process sequences of variable length.
            - Backpropagation Through Time (BPTT): This is the training algorithm used for RNNs. It calculates gradients by unrolling the network and summing errors across all time steps. 
         - Design Criterias that RNNs meet in order to model sequence data:
            1. Handle variable length sequences
            2. Track long-term dependencies
            3. Maintain information about order
            4. Share parameters across the sequence
