## In Class Notes

- Classification Problem: The values (y) which we want to predict are usually discrete variables.
    - Binary Classification: The values (y) which we want to predict are restricted to only two choices (Classes/labels). Usually is denoted by a 1 or 0, where one is True or positive class and 0 is False or negative class.
    - Multiclass Classification: The values (y) which we want to predict can belong to one of many discrete values. The number of classes in these cases are usually greater than 0, but have a fixed range.
- Logistic Regression: Logistic Regression is a supervised machine learning algorithm primarily used for classification tasks, particularly binary classification, where the outcome is categorical (e.g., yes/no, 0/1, malignant/benign). Unlike Linear Regression, which predicts continuous values, Logistic Regression predicts the probability of an event occurring.
    - The goal of Logistic Regression is to estimate the probability of occurrence.
    - Logistic Regression is designed to predict the probability of a given outcome.
    - Unlike linear regression, whose outputs can span between $ -\infty  and  \infty$. Logistic Regression, constrains predictions between 0 and 1 using the sigmoid (logistic) function:

    - $ \sigma(t) = \frac{1}{1+e^-t} , where t = \beta_0 + \beta_1 x $
    - Logistic Regression uses the Logistic Function.
    - The following equation yields valid probabilities of a data point (t) belonging to 0 or 1 and is called the logistic function or the sigmoid function. Here is a plot showing $ \sigma(t) $
    - 
    - Note that, $ \sigma(t) $ tends to 1 as t reaches $ \infty $ and $ \sigma(t) $ tends to 0 as t reaches $ -\infty $
    - Since $ t = \beta_0 + \beta_1 x $ we need to optimize the slobe and the intercept to make sure that the outcomes are generalized and accurate to the data.
    - Logistic regression uses Maximum Likelihood Estimation (MLE): choose $\beta$ parameters that maximize the likelihood of observing actual outcomes.

    - Sigmoid Function: We typically denote the sigmoid function by the greek letter $\sigma$(sigma) and define as
        - $ \sigma(x)  = \frac{1}{1 + e^-x}$
            - x is the input and $ e $ is the euler's number which approximates to 2.781
        - The derivative of the sigmoid function is:
            - $ \frac{d}{dx}\sigma(x) = \sigma(x)(1 - \sigma(x)) $
            - Significance of the derivative of the Sigmoid Function: calculating the derivative of the sigmoid function is very important, as it tells us how to optimize the parameters of our model with gradient descent to improve performance.
            - This expression of the derivative is very convenient since, in most use cases, we have already calculated $\sigma(x)$ in our model before attempting gradient descent (e.g. during the feedforward step in neural networks). If we want to apply gradient descent now, we can insert $\sigma(x)$ into two spots of the derivative equation without any high calculation costs.

- Issues with the Sigmoid Function:
    -  The gradient vanishes to zero for very low and very high input values, making it hard for some models to improve.
    - For example, during backpropagation in deep learning, the gradient of a sigmoid activation function is used to update the weights & biases of a neural network. If these gradients are tiny, the updates to the weights & biases are tiny and the network will not learn.
    - Alternatively, other non-linear functions such as the Rectified Linear Unit (ReLu) are used, which do not show these flaws.
