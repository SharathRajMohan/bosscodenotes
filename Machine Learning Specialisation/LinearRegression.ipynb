# In Class Notes
- Linear Regression:
    - Linear regression models the relationship between a dependent variable **Y** and one or more independent variables $X_1, X_2, …, X_k$. 
    - The **simple linear regression** formula is:
        - $ \hat{y} = \beta_0 + \beta_1 x $
    - For multiple linear regression:
        - $ \hat{y} = \beta_0 + \beta_1 x_1 + + \beta_2 x_2 + ... + \beta_k x_k$
    - Here, 
        - $ \beta_0 $ : Is the intercept.
        - $ \beta_j $ : Are the coefficients. How much 'y' changes for a unit change in each 'X'.
    - Model fitting is typically done by Ordinary Least Squares (OLS), which minimizes the sum of squared residuals (differences between actual and predicted values). This is an iterative process, repeated until an acceptable equation is derived, which yields an average minimized error.
- Ordinary Least Squares: is the most common technique used to estimate the parameters (coefficients) of a linear regression model.
- In regression problems the target variable is continuous in nature.
- Dependent variable: The outcome or result we are trying to predict. This is usually continuous in nature for regression problems.
- Independent variable: The variables that influence the dependent variable is known as independent variable.
- Essentially, the dependent variable 'depends' on the independent variable.
- Assumptions of Machine learning models
    - Linearity: There must exist an independent linear relationship between the dependent variable and the independent variable.
        - 
        - Can check this by use of visualization. (Scatterplot between x and y)
        - Check co-linearity. (Co-relation heatmap)
    - Homoscedasticity (Equal Variance): 
        - 
        - Means that the variance must be equal or constant throughout the set of predictions.
        - The variance of the model must not be dependent on the variance of the independent variable.
        - In practical terms, this means that the spread of the residuals should remain approximately the same as the predicted values increase.
        - The violation of this assumption results in heteroscedasticity.
        - If the scatter plot displays a funnel-shaped pattern or if the spread of residuals widens or narrows systematically with the predicted values, it suggests heteroscedasticity, which violates the homoscedasticity assumption.
        - Check whether the dispersion of residuals remains relatively constant across the range of predicted values. In other words, the variability of residuals should not systematically change as the predicted values increase or decrease.
    - Multivariate Normality (Normality of error distributions)
        - 
        - The errors (residuals) of the model are normally distributed.
        - While this assumption is not strictly necessary for large sample sizes (thanks to the Central Limit Theorem --> as the distribution tends to become normal with the increase in sample size), it can be important for smaller sample sizes to ensure the validity of statistical tests and confidence intervals.
        - In cases where the distribution of the residuals are not normal we have to apply transformations to normalize the data.
    - Independence of Errors (No Auto-corelation): The residuals (actual - predicted) must not be co-related to each other
        - 
        - Auto co-relation is also known as serial correlation: means that the error terms in your regression model are not independent, and the value of an error term is related to the values of previous error terms. 
        - The violation of the rule indicates that the data are co-related to each other and follow some sort of pattern or trends.
        - Avoid trends and patterns in the data.
        - Consequences of Autocorrelation:
            -  When autocorrelation is present, the standard errors of the regression coefficients can be underestimated, leading to inflated t-statistics and potentially incorrect conclusions about the significance of the variables.
            - This violates the assumption of independent errors and can lead to biased parameter estimates and incorrect statistical inference.
    - Lack of Multicollinearity
        - 
        - In multiple linear regression (where there are more than one independent variable), there should be no perfect linear relationship between the independent variables.
        - Perfect multicollinearity occurs when one independent variable can be exactly predicted from another independent variable or a combination of other independent variables.
        - This situation makes it impossible to estimate the unique effect of each independent variable on the dependent variable.
        - If we proceed to build a model in such situations the co-efficient(s) we obtain would be unreliable.
        - Predictors are not correlated with each other

- Error (LOSS): The difference between the actual value and the predicted value. AKA cost
    - $Error = Actual\ Value - Predicted\ Value$
- Evaluation Metrics
    - Types of LOSS:
        - L1 Loss: The sum of the absolute values of the difference between the predicted values and the actual values.
            - $\sum|actual - predicted|$
        - L2 Loss: The sum of the squared difference between the predicted values and the actual values.
            - $\sum(actual - predicted)^2$
        
    - $$MSE = \frac{1}{n} \sum_{i=1}^{n}(y_i-\hat{y_i})^2$$
        - Captures average squared errors—penalizes larger deviations more heavily.
        - Used both as a loss function in OLS and as a performance metric.
        - Note: This metrics punishes outliers or extreme values as they get squared, hence may cause the error to be inflated.

    - RMSE
        - $$RMSE = \sqrt{MSE}$$
        - Returns error in original units, making interpretation easier.
        - Still strongly penalizes outliers.
    - MAE
        - $$ MAE = \frac{1}{n} \sum_{i=1}^{n}|y_i - \hat{y_i}|$$
        - Averages absolute errors—less sensitive to outliers.
        - Easier to interpret, linear penalty, same unit as errors.
    - MAPE
        - $$ MAPE = \frac{100}{n} \sum_{i=1}^{n}|\frac{y_i - \hat{y_i}}{y_i}|$$
        - Expresses error in percentages—scale-independent and intuitive.
        - Con: Distorts when actual values are near zero; could over-penalize small denominators.
    - $R^2$ R-squared or Coefficient of determination
        - This metric represents the part of the variance of the dependent variable explained by the independent variables of the model. It measures the strength of the relationship between your model and the dependent variable.
- Best practices while evaluating a model using it's metrics:
    - Compute multiple metrics to get a complete error profile.
    - Compare against a baseline (e.g., predicting the mean).
    - Visualize residual distributions and patterns alongside metrics.
- Cost Function
- Best Fit Line: The line that yields the least consolidated residuals.
- Methods for computing the best fit line (coefficients $\beta$)
    - Ordinary Least Squares: Ordinary Least Squares (OLS) is the most common technique used to estimate the parameters (coefficients) of a linear regression model.
        - OLS estimates the coefficients $\beta$ in the regression equation
            - $ \hat{y} = \beta_0 + \beta_1 x_1 + + \beta_2 x_2 + ... + \beta_k x_k$
            - by minimizing the sum of squared residuals
        - Under the classical linear regression assumptions (linear model, exogeneity, homoscedasticity, no autocorrelation), the Gauss–Markov theorem guarantees that OLS provides the lowest variance among all unbiased linear estimators
        - As the sample size grows, OLS coefficients converge to the true parameter values—if regressors are exogenous (no correlation with error term)
        - When errors are also normally distributed, OLS becomes the maximum likelihood estimator. This allows for precise hypothesis testing, confidence intervals, and R²-based model evaluation
        - OLS is simple to compute, interpret, and apply across fields like economics (wage/income analysis), finance (CAPM beta estimation), social sciences, and engineering
        - When OLS struggles:
            - When there are presence of outliers in the data.
            - Violation of assumptions: Issues like heteroscedasticity affect standard errors and inference, while multicollinearity muddles coefficient estimates .
    - Gradient Descent: Gradient descent is an iterative optimization algorithm designed to minimize a cost function—typically the Mean Squared Error (MSE) in linear regression.
        - is a mathematical technique that iteratively finds the weights and bias that produce the model with the lowest loss.
        - Since, it's an iterative process the number of iterations are preset by the user.
        - Process behind Gradient Descent:
            - The model begins training with randomized weights and biases near zero, and then repeats the following steps:
                1. Calculate the loss (cost) with the current weight and bias. [COST FUNCTION]
                    - $ J(m,b) = \frac{1}{n} \sum_{i=1}^{n} (y_i - (mx_i + b))^2 $
                    - Our goal is to minimize J(m,b) with respect to the parameters m and b.
                2. Determine the direction to move the weights and bias that reduce loss. [COMPUTE GRADIENTS]
                    - Calculate the slope of the tangent to the loss function at each weight and the bias
                    - [Math behind Gradient Descent](#ml-gd)
                    - The gradient of the weight is denoted by $\frac{\partial(J)}{\partial(m)}$
                    - The gradient of the bias is denoted by $\frac{\partial(J)}{\partial(b)}$
                3. Move the weight and bias values a small amount in the direction that reduces loss. [UPDATE RULE]
                    - Learning rate ($\alpha$): Used as a factor to determine by what rate the parameters (m,b) must change.
                    - the learning rate 𝛼 *scales* how far you move in that direction each iteration
                    - The learning rate (α) is crucial. A high learning rate may cause overshooting, bounce back and forth or diverge(loss blows up) while a low learning rate may slow down convergence. Can also get stuck in a suboptimal point (local minima).
                    - Update rules apply the gradient in small steps controlled by the learning rate $\alpha$
                    - The $\alpha$ Effect:
                        - Loss oscillating or exploding → α is likely too high.
                        - Loss decreasing very slowly → α is likely too low.
                        - Converges but plateaus early → consider α schedule or smaller final learning rate for fine-tuning.
                    - Use grid search and loss monitoring to find the best α for your model.
                    - Feature scaling (normalization/standardization) is also key—it helps ensure gradients have consistent magnitudes across features, allowing α to be set more effectively .
                    - Setting the new weights and bias is denoted by:
                        - $ m(new) =  m(old) - \alpha \frac{\partial(J)}{\partial(m)}$
                        - $ b(new) = b(old) - \alpha \frac{\partial(J)}{\partial(b)}$
                    - Alternative explanation for the math behind the parameter update process.
                        - 
                4. Return to step one and repeat the process until the model can't reduce the loss any further. [CONVERGENCE]

- Maximum Likelihood Estimation (MLE): is a method for estimating the parameters of a probabilistic model by finding the values that make the observed data most probable.
    - Used to find the optimal value for the intercept and slope of the line that maximizes the likelihood of it being the best fit to the data.
    - It typically, places a normal distribution curve around the line, the distribution we use must reflect the data points around the line.
    - 
    - The following formula depicts the normal distribution function:
        - $$f(y) = \frac{1}{\sqrt{2\pi\sigma^2}}e^-\frac{(y-\mu)^2}{2\sigma^2}$$
            - The 2 influential parameters are 
                - $\mu$ or the mean
                - $\sigma$ or the standard deviation
            - Since the curves are centered along the lines, the value for:
                - $\mu$ = corresponds to the y value of the line, and can be replaced with the equation for the line.
                - $\sigma$ = corresponds to the width of the curve and is estimated based on the spread of the data points around the line.

- Feedback: Refers to the process of using the information provided by the cost function to adjust the model parameters in order to minimize the error.
- Loss Curves:Graphical representations (over iterations or epochs) showing how the loss (e.g., MSE) decreases during training. They help visualize model performance, detect convergence, and diagnose issues like overfitting or underfitting .
- Model Convergence
    - When a model converges, additional iterations don't reduce loss more because gradient descent has found the weights and bias that nearly minimize the loss.
    - Convex Functions
        - For linear regression, the cost function is convex (no local minima), so gradient descent reliably finds the global minimum
    - If the model continues to train past convergence, loss begins to fluctuate in small amounts as the model continually updates the parameters around their lowest values. This can make it hard to verify that the model has actually converged. To confirm the model has converged, you'll want to continue training until the loss has stabilized.
- Bias
    - Error due to overly simplistic assumptions in the model. It's the difference between the average prediction across training sets and the true value.
    - High bias indicates underfitting, where the model misses important patterns and performs poorly on both training and test data
- Variance
    - Error due to sensitivity to fluctuations in the training data. It's the variability of model predictions when trained on different data samples.
    - High variance leads to overfitting—strong performance on training data but poor generalization to new data
- Bias Variance Tradeoff:
    - Principle: Total expected error decomposes into bias² + variance + irreducible error.
    - Tradeoff: Reducing one often increases the other—you can’t minimize both simultaneously. The goal is to find an optimal balance to achieve low prediction error on unseen data.
    - Implication:
        - High bias → underfitting
        - High variance → overfitting
        - Optimal → balanced low bias and low variance

### Math behind GD

- To get the slope for the lines tangent to the weight and bias, we take the derivative of the loss function with respect to the weight and the bias, and then solve the equations.
- We know, that the equation for making the prediction is:
    - $ F(x) = m(x) + b $
        - Where m --> slope(weight) and b --> intercept(bias)
    - And, the cost is calculated for 'n' samples as 
        - $ J(m,b) = \frac{1}{n} \sum_{i=1}^{n} (y_i - (mx_i + b))^2 $
            - Where $y_i$ is the actual prediction.   
    
- Weight Derivative ($\frac{\partial(J)}{\partial(m)}$)
    - $ \frac{\partial(J)}{\partial(m)} = \frac{1}{n} \sum_{i=1}^{n} (y_i - (mx_i + b)) * 2x_i $
    - First we sum each predicted value minus the actual value and then multiply it by two times the feature value. Then we divide the sum by the number of examples. The result is the slope of the line tangent to the value of the weight.
    - $\frac{\partial(J)}{\partial(m)}$ --> Shows how adjusting the slope changes the error, weighted by $x_i$.

- Bias Derivative ($\frac{\partial(J)}{\partial(b)}$)
    - $ \frac{\partial(J)}{\partial(b)} = \frac{1}{n} \sum_{i=1}^{n} (y_i - (mx_i + b)) * 2 $
    - First we sum each predicted value minus the actual value and then multiply it by two. Then we divide the sum by the number of examples. The result is the slope of the line tangent to the value of the bias.
    - $\frac{\partial(J)}{\partial(b)}$ -> Shows how adjusting the intercept affects the error uniformly across the data points.

- The negative signs indicate we move parameters against the gradient to reduce the loss.

- Additional Notes:
    - These come from applying the chain rule:
        1. The derivative of the squared error term:
            - $ \frac{d}{du}(u^2) = 2u$
        2. This gradient points in the steepest ascent direction, so we move opposite it to descend toward the minimum.

## 🏋️ 1. Batch Gradient Descent (BGD)

* **Update rule**: Compute gradient using the **entire dataset** each iteration.
* **Pros**: Stable convergence, smooth update directions toward the minimum.
* **Cons**: Memory-heavy and slow for large datasets.
* **Best used for**: Smaller datasets with convex loss surfaces.

---

## 🎢 2. Stochastic Gradient Descent (SGD)

* **Update rule**: Compute gradient using a **single randomly selected sample** each step.
* **Pros**: Fast updates, low memory footprint, can escape local minima due to noise.
* **Cons**: Noisy convergence, requires careful learning rate tuning.
* **Best used for**: Large datasets or streaming/online learning.

---

## ⚖️ 3. Mini-Batch Gradient Descent (MBGD)

* **Update rule**: Compute gradient over a **small batch** (e.g., 32–512 samples).
* **Pros**: Smooths noise from SGD, efficient with hardware, good trade-off between BGD and SGD.
* **Cons**: Extra tuning for batch size; still noisy vs full BGD.
* **Best used for**: Deep learning, large datasets, hardware acceleration.

---

## 🚀 4. Momentum & Accelerated Methods (e.g., Nesterov)

* Add a **velocity term** to incorporate past gradients for smoother, faster descent and to escape shallow region.
* **Nesterov acceleration** further improves convergence rates to $O(1/k^2)$ for convex problems .

---

## 🔄 5. Adaptive Learning Rate Methods

* **AdaGrad**: Scales learning rate per parameter from past gradients.
* **RMSprop** & **Adam**: Adjust rates via moving averages of gradients and squared gradients; commonly used in deep learning.
* **Pros**: Automatically tuned steps, improved stability.
* **Cons**: May generalize worse; adds complexity in tuning β₁/β₂.

---

## 📉 6. Variance Reduction Techniques (e.g., SVRG)

* Use **periodic full-gradient estimates** to adjust stochastic gradients—reducing noise and improving convergence.

---

## 🪞 7. Mirror Descent

* Generalizes gradient descent using **Bregman divergence**, suited for optimization in **non-Euclidean spaces**.

### 💡 Key Insights

* **BGD** is precise but impractical with big data.
* **SGD** offers speed and memory efficiency but is noisy.
* **Mini-batch** strikes a productive balance and is widely adopted.
* **Momentum/NAG** boost convergence and stability.
* **Adaptive methods** like Adam are effective in deep learning, though they can affect generalization.
* **Variance reduction** ensures faster convergence with fewer iterations.
* **Mirror descent** extends gradient methods to different mathematical frameworks.
