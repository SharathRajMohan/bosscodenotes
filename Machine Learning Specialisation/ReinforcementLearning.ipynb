# Reinforcement Learning RL

### Chapter 1: 
- In reinforcement learning (RL) there is:
    - Agent: The entity whose actions you can control.
    - Environment: The entity that you have no control of, but can interact with using the agent. The agent is part of the environment.
- Influence of the Agent on the Environment:
    - Action: The influence of the agent on the environment.
    - State: The influence of the environment on the agent.
    - Reward: An indicator that let's the agent know if the action had a good or bad outcome.
    - 
- In RL, the problem that we wish to solve needs to have a structure this makes it easier to solve. RL frames a task in a certain manner which makes it easier to learn and solve known as Markov Decision Process.
- Markov Decision Process:
    - It's a discreet, repeating ordered sequence of processes. Each process consists of a State(S), Action(A) and a Reward(R).
    - $$ S_{0} , A_{0}, R_{0},S_{1} , A_{1}, R_{1},S_{2} , A_{2}, R_{2}...S_{n} , A_{n}, R_{n}$$
    - The MDP can also continue unrestricted for a number of iterations or be confined to a number of iterations, this is known as an Episode.
    - Markov Property: Each State(S) is only dependant on it's immediately previous state.
    - Within an MDP, the ultimate goal of the agent is to figure out how to use an input State(S) to decide on an output Action(A), in a way to maximize the total subsequent Reward(R) over time.
- Policy($\pi$): A function that takes State(S) as an input and gives an Action(A) as an output. Over time this has evolved to output the probability of choosing a specific action from a list of actions. It would output which is the most likely action to be performed based on the state. represented as :
    - $$ \pi(A|S) => Probability(A) $$
- Return($G_t$): The total cumulative reward upto the time(t).
- GOAL: Find the Policy($\pi$) which maximizes the Return($G_t$).
- World Model: Model of the environment. Simply means that the agent is aware of the properties of the environment and performs actions accordingly to maximize rewards.
    - Mathematically, this is given as:
        - P:(s,a) -> s',r
            - ie, Given that you picked an Action(a) for a certain state (s) what is the probability of the next state and reward.
        - P(s',r|s,a)
